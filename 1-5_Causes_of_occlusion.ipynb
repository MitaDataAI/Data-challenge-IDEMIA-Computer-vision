# Conclusion
- Causes: hair, hands, accessories (e.g., glasses, masks), shadows, and camera angles.
- Low occlusion (< 0.2): ~80% of images — ideal for recognition.
- Moderate occlusion (0.2–0.4): ~13% — common in real life.
- High occlusion (> 0.4): <1% — rare but challenging.
- Occlusion reflects real-world behavior, not noise.
- Models should be trained with diverse occlusion levels to ensure robustness.

## FaceOcclusion between 0.0 and 0.1 (64,043 images)
Faces in this range are almost entirely unobstructed. This category serves as the ideal input for facial recognition systems. They ensure maximum accuracy under optimal conditions.

## FaceOcclusion between 0.1 and 0.2 (20,545 images)
Slight occlusions begin to appear in this group—such as eyeglasses, light shadows, or minor hair interference. While facial features remain mostly visible, these small obstructions may introduce subtle challenges for recognition algorithms.

## FaceOcclusion between 0.2 and 0.3 (9,242 images)
Moderate occlusions are more evident here, including partially covered faces due to accessories, hair, or hands. These images represent more realistic scenarios where occlusions are common.

## FaceOcclusion between 0.3 and 0.4 (3,391 images)
Faces in this group experience noticeable obstruction—such as objects blocking parts of the face or side poses combined with occlusion. Recognition and detection become significantly more difficult.
